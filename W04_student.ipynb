## valid_count (5 marks)

## mean_reading (5 marks)
Store the arithmetic mean of the nonblank crack_width_mm values as a number. Add the valid readings and divide by their count. Keep full precision.


```python
# Your calculation here
import csv

with open("site_inspection_log.csv", newline="") as f:
    rows = list(csv.DictReader(f))

valid_readings = []
for r in rows:
    raw = r["crack_width_mm"].strip()
    if raw == "":
        continue
    valid_readings.append(float(raw))

mean_reading = sum(valid_readings) / len(valid_readings)
print(mean_reading)
```

## above_limit (5 marks)
Store an integer count of valid crack_width_mm readings strictly greater than 0.3. A reading equal to the threshold is not counted.


```python
# Your calculation here
import csv

with open("site_inspection_log.csv", newline="") as f:
    rows = list(csv.DictReader(f))

valid_readings = []
for r in rows:
    raw = r["crack_width_mm"].strip()
    if raw == "":
        continue
    valid_readings.append(float(raw))

above_limit = sum(1 for x in valid_readings if x > 0.3)
print(above_limit)
```

## group_means (5 marks)
Store a dictionary whose keys are the structure_id labels and whose values are the mean valid crack_width_mm readings within each group. Divide each group total by its own valid count. Keep full precision.


```python
# Your calculation here
import csv

with open("site_inspection_log.csv", newline="") as f:
    rows = list(csv.DictReader(f))

totals = {}
counts = {}

for r in rows:
    raw = r["crack_width_mm"].strip()
    if raw == "":
        continue
    key = r["structure_id"]
    totals[key] = totals.get(key, 0) + float(raw)
    counts[key] = counts.get(key, 0) + 1

group_means = {key: totals[key] / counts[key] for key in totals}
print(group_means)
```
